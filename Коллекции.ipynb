```python
#1.
s = input("Введите строку: ")

result = {}
for ch in s:
    if ch in result:
        result[ch] += 1
    else:
        result[ch] = 1

print(result)
```

```python
#2.
s = input("Введите текст: ")

words = s.split()
reversed_words = []
for i in range(len(words) - 1, -1, -1):
    reversed_words.append(words[i])

result = " ".join(reversed_words)
print(result)
```

```python
#3.
data = input("Введите элементы через пробел: ").split()

unique = []
for item in data:
    if item not in unique:
        unique.append(item)

print(unique)
```

```python
#4.
a = input("Список 1 (через пробел): ").split()
b = input("Список 2 (через пробел): ").split()
c = input("Список 3 (через пробел): ").split()

merged = []
for lst in [a, b, c]:
    for item in lst:
        if item not in merged:
            merged.append(item)

print(merged)
```

```python
#5.
text = input("Введите текст: ")

vowels = "аеёиоуыэюяaeiouy"
consonants = "бвгджзйклмнпрстфхцчшщbcdfghjklmnpqrstvwxyz"
digits = "0123456789"
punctuation = ".,;:!?\"'()[]{}<>-—–/@#№$%^&*_+=|\\~`"

count = {"гласные": 0, "согласные": 0, "цифры": 0, "пунктуация": 0}

for ch in text.lower():
    if ch in vowels:
        count["гласные"] += 1
    elif ch in consonants:
        count["согласные"] += 1
    elif ch in digits:
        count["цифры"] += 1
    elif ch in punctuation:
        count["пунктуация"] += 1

print(count)
```
