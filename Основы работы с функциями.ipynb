```python
#1.
def calculate(operation, *numbers):
    if operation == "add":
        result = 0
        for n in numbers:
            result += n
        return result
    elif operation == "mul":
        result = 1
        for n in numbers:
            result *= n
        return result
    else:
        return "Неизвестная операция"
print(calculate("add", 1, 2, 3, 4))    # 10
print(calculate("mul", 2, 3, 4))       # 24
```

```python
#2.
def input_int(prompt="Введите целое число: "):
    while True:
        value = input(prompt)
        try:
            return int(value)
        except ValueError:
            print(f"'{value}' — это не целое число. Попробуйте снова.")
age = input_int("Сколько вам лет? ")
print(f"Вы ввели: {age}")
```

```python
#3.
def full_path(drive, *folders, filename, ext=None, sep="/"):
    parts = [drive.rstrip(":/\\")]
    parts.extend(folders)
    parts.append(filename)
    path = sep.join(parts)
    if ext:
        path += "." + ext.lstrip(".")
    return path
print(full_path("c:", "work", "python", "function", "main", filename="main", ext="py"))
print(full_path("d:", "data", filename="report", ext="txt", sep="\\"))
print(full_path("c:", filename="readme"))
```

```python
#4.
def group_by_type(items):
    result = {}
    for item in items:
        t = type(item)
        if t not in result:
            result[t] = []
        result[t].append(item)
    return result
data = [1, "hello", 3.14, True, 42, "world", 2.5, False]
grouped = group_by_type(data)
for t, values in grouped.items():
    print(t.__name__, "→", values)
```
