```python
import numpy as np

A = np.array([[2, -4],
              [3, 5],
              [-1, 0]])

B = np.array([[1, 2, 7],
              [-3, -4, 0],
              [5, 2, 1]])

C = np.array([[6, -3, 9],
              [4, -5, 2],
              [8, 1, 5]])

D = A.T @ C - 2 * A.T @ B.T
print(D)
```

```python
import numpy as np

x = np.array([-4, 2, 4])
y = np.array([2, 2, -3])

# Задание 3: сумма |y - kx|
ks = np.linspace(-2, 2, 4001)
F = np.array([np.sum(np.abs(y - k * x)) for k in ks])
print("k (модули):", ks[F.argmin()], "F_min =", F.min())

# Задание 4: аналитически k = sum(x*y) / sum(x^2)
k_opt = np.sum(x * y) / np.sum(x * x)
print("k (МНК):", k_opt)
print("Q_min =", np.sum((y - k_opt * x) ** 2))
```

```python
import matplotlib.pyplot as plt
import numpy as np

x_pts = [-4, 2, 4]
y_pts = [2, 2, -3]

fig, ax = plt.subplots(figsize=(7, 7))

xs = np.linspace(-6, 6, 200)
ax.plot(xs, -0.5 * xs, label="y = -0.5x (сумма |отклонений|)", color="blue")
ax.plot(xs, -4/9 * xs, label="y = -4/9x (сумма квадратов)", color="red", linestyle="--")

ax.scatter(x_pts, y_pts, color="black", zorder=5, s=80)

ax.axhline(0, color="grey", linewidth=0.5)
ax.axvline(0, color="grey", linewidth=0.5)
ax.grid(True)
ax.set_aspect("equal")
ax.legend()
plt.title("Прямые через (0,0), минимизирующие отклонения")
plt.show()
```

```python
import numpy as np

A = np.array([[1, 2, 4],
              [2, 1, -1],
              [1, -1, 1]])

b = np.array([3, -9, 0])

x = np.linalg.solve(A, b)
print(x)         # [-3. -1.  2.]

print(A @ x)     # [ 3. -9.  0.] — совпадает с b ✅
```

```python
import sympy as sp

x = sp.Symbol('x', real=True)

# Плотность
pdf = 2 * x

# 6-1. Вероятность
p = sp.integrate(pdf, (x, 0, sp.Rational(1, 2)))
print("P(0 <= X <= 0.5) =", p)   # 1/4

# 6-2. CDF на [0, 1]
cdf = sp.integrate(pdf, (x, 0, x))
print("F(x) =", cdf)             # x**2
```
